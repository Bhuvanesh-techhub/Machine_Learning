```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression 
from sklearn.metrics import accuracy_score

#1.Loading Dataset
iris = load_iris()
X = iris.data
y = iris.target
#2.Splitting the Dataset
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state = 1)

#3.Training the Model
log_reg = LogisticRegression(max_iter=200) #Creating a logisitc regression classifier object\
log_reg.fit(X_train, y_train)#Training the model with the training set

#4.Making Predictions
y_pred = log_reg.predict(X_test)#Making predicitons on the test data

#5.Evaluating Model Accuracy
accuracy = accuracy_score(y_test,y_pred) #Comparing the predicted labels to the actual test labels
print(f"Logistic Regression Model Accuracy:{accuracy}")

# Make predictions on new sample data (out-of-sample data)
new_data = [[5.0, 3.6, 1.3, 0.25], [6.5, 3.0, 5.2, 2.0]] # Example new data
predictions = log_reg.predict(new_data)
print(f"Predictions for new data: {iris.target_names[predictions]}")
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Loading Dataset
iris = load_iris()
X = iris.data
y = iris.target

# 2. Splitting the Dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Training the Model
log_reg = LogisticRegression(max_iter=200)
log_reg.fit(X_train, y_train)

# 4. Making Predictions
y_pred = log_reg.predict(X_test)

# 5. Evaluating Model Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Logistic Regression Model Accuracy: {accuracy:.2f}")

# 6. Predictions on new sample data
new_data = [[6.0, 2.6, 5.3, 2.25], [6.5, 3.0, 5.2, 2.0]]
predictions = log_reg.predict(new_data)

# Map predictions to flower names
predicted_names = [iris.target_names[p] for p in predictions]
print(f"Predictions for new data: {predicted_names}")
```
